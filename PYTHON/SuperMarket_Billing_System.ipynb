```python
from datetime import date
import random

# ----------- HEADER -----------
print("=" * 75)
print("              SUPER MARKET BILLING SYSTEM")
print("=" * 75)
print("date:", date.today())
print("bill no:", random.randint(100, 1000))
print("=" * 75)

# ----------- DATA -----------
categories = {
    "grocery": {
        "rice": 50,
        "wheat": 45,
        "sugar": 40,
        "oil": 120,
        "dal": 90
    },
    "vegetable": {
        "potato": 30,
        "onion": 35,
        "tomato": 25,
        "carrot": 50,
        "capsicum": 20
    },
    "skincare": {
        "facewash": {
            "mamaearth": 250,
            "m-caffine": 300
        },
        "moisturizer": {
            "nivea": 220,
            "dot & key": 280
        },
        "sunscreen": {
            "minimalist": 350,
            "aqualogica": 300
        }
    }
}

discount_rate = {
    "grocery": 0.10,
    "vegetable": 0.05,
    "skincare": 0.15
}

# ----------- BILL STORAGE -----------
bill_items = []
subtotal = 0
total_discount = 0

# ----------- INPUT LOOP -----------
while True:
    print("\ncategories:", list(categories.keys()))
    category = input("enter category: ").strip().lower()

    if category not in categories:
        print("invalid category")
        continue

    # grocery / vegetable
    if category in ["grocery", "vegetable"]:
        print("available items:", list(categories[category].keys()))
        item = input("enter item name: ").strip().lower()

        if item not in categories[category]:
            print("invalid item")
            continue

        price = categories[category][item]
        qty = int(input("enter quantity: "))

        item_total = price * qty
        discount = item_total * discount_rate[category]
        final_price = item_total - discount

        bill_items.append([item, qty, price, final_price])

    # skincare
    else:
        print("products:", list(categories["skincare"].keys()))
        product = input("enter product type: ").strip().lower()

        if product not in categories["skincare"]:
            print("invalid product")
            continue

        print("brands:", list(categories["skincare"][product].keys()))
        brand = input("enter brand: ").strip().lower()

        if brand not in categories["skincare"][product]:
            print("invalid brand")
            continue

        price = categories["skincare"][product][brand]
        qty = int(input("enter quantity: "))

        item_total = price * qty
        discount = item_total * discount_rate["skincare"]
        final_price = item_total - discount

        item_name = f"{product} ({brand})"
        bill_items.append([item_name, qty, price, final_price])

    subtotal += item_total
    total_discount += discount

    more = input("add more items? (yes/no): ").strip().lower()
    if more != "yes":
        break

# ----------- FINAL BILL -----------
print("\n" + "-" * 75)
print(f"{'item name':<25}{'qty':<8}{'price':<10}{'total':<10}")
print("-" * 75)

for item in bill_items:
    print(f"{item[0]:<25}{item[1]:<8}{item[2]:<10}{item[3]:<10.2f}")

print("-" * 75)
print(f"{'subtotal:':<43}{subtotal:.2f}")
print(f"{'discount:':<43}{total_discount:.2f}")
print("-" * 75)
print(f"{'final amount:':<43}{subtotal - total_discount:.2f}")
print("-" * 75)
print("thank you for shopping with us!")
print("=" * 75)
```
